```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Excess Mode dataset
data = pd.read_csv("/content/Hybrid_Excess_Condition.csv")

# Display first few rows
print(data.head())

# Plot Fuel Cell Power
plt.figure()
plt.plot(data['FuelCell'])
plt.title("Fuel Cell Power (Excess Mode)")
plt.xlabel("Samples")
plt.ylabel("Power")
plt.show()

# Plot Battery Power (Charging in excess mode)
plt.figure()
plt.plot(data['Battery'])
plt.title("Battery Power (Charging Mode)")
plt.xlabel("Samples")
plt.ylabel("Power")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("/content/Hybrid_Deficit_Condition.csv")

# Print columns to verify
print(data.columns)

# Example: change names based on your dataset
fuel = data.iloc[:, 0]   # first column (Fuel Cell)
battery = data.iloc[:, 1]  # second column (Battery)

# Plot Fuel Cell Power
plt.figure()
plt.plot(fuel)
plt.title("Fuel Cell Power (Excess Mode)")
plt.show()

# Plot Battery Power
plt.figure()
plt.plot(battery)
plt.title("Battery Charging (Excess Mode)")
plt.show()
```
